```python
def Two_sum(nums,target):

    n = len(nums)

    hash_map = {}

    for i in range(n):

        current = nums[i]

        remaining_value = target - current

        if remaining_value in hash_map:
            return [hash_map[remaining_value],i]
        
        hash_map[current] = i

nums = nums = [2,7,11,15]
target = 26
result = Two_sum(nums,target)
print(result)
```
