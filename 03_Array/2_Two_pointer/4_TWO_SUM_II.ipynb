## __Leetcode Two Sum - II__

```python
## Code 
## Optimal Approach
## Two Pointer Pattern

def Two_sum_II(nums,target):

    ## Len of Nums
    n = len(nums)

    #Pointers
    left = 0
    right = n - 1

    ## Loop To Traverse the Whole Array
    while left <= right:

        # if the left + right == target : return 
        if nums[left] + nums[right] == target:
            
            # Python uses 0-based indexing, but LeetCode 167 (Two Sum II) asks you to return 1-based positions.
            return [left + 1, right + 1]

        elif nums[left] + nums[right] < target:

            left = left + 1

        else:

            right = right - 1


nums = [2,7,8,33,35]
target = 15

result = Two_sum_II(nums,target)

print(result)


## Time Complexity = O(n) - It traverse the n Time whole nums/array.
## Space Complexity = O(1) - No extra Space is usingg
```
