```python
## Brute Force Approach is we can append and store in the list..
# But Question Say we have to in - place so we cant make new list..\/
#Time Complexity = O(n)
#Space Complexity = O(1)
## Optimal Approach / Better Solution / My - Try

def sort_Colours(nums):
    
    # Len of Nums
    n = len(nums)

    #Two Pointer Where There starting Position
    left = 0
    mid = 0
    right = n - 1

    while mid <= right:

        #Condition
        if nums[mid] == 0:

            nums[left],nums[mid] = nums[mid] , nums[left]

            left = left + 1
            mid = mid + 1

        elif nums[mid] == 1:

            mid = mid + 1


        elif nums[mid] == 2:
            nums[mid],nums[right] = nums[right] , nums[mid]
            right = right - 1

           


    return nums


#Driver Code
nums = [2,0,1,1,0,2]

#Function Calling
result = sort_Colours(nums)
print("After sort the colours the nums is :-",result)
```

```python
def sort_Colours(nums):

    # Three pointers
    left = 0
    mid = 0
    right = len(nums) - 1

    # Process elements until mid crosses right
    while mid <= right:

        # If current element is 0
        if nums[mid] == 0:

            # Move 0 to the left side
            nums[left], nums[mid] = nums[mid], nums[left]

            left += 1
            mid += 1

        # If current element is 1
        elif nums[mid] == 1:

            # 1 is already in the middle
            mid += 1

        # If current element is 2
        else:

            # Move 2 to the right side
            nums[mid], nums[right] = nums[right], nums[mid]

            right -= 1

            # Do NOT increase mid
            # because the new element at mid is not checked yet

    return nums


# Driver Code

nums = [2, 0, 2, 1, 1, 0]

result = sort_Colours(nums)

print("After sorting the colours, nums is:", result)
```

## The Time Complexity = O(n) - we traverse array only n time
## Space Complexity = O(1) - Because we did not use any extra space
