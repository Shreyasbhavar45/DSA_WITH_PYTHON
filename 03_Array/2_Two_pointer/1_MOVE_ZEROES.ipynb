## Q7 — __Move Zeroes (LeetCode 283)x__

```python
## Optimal Force Approach


def moveZeroes(nums):

    # Len of array
    n = len(nums)

    left = 0

    for right in range(n):

        if nums[right] != 0:

            nums[left],nums[right] = nums[right],nums[left]


            left = left + 1

    return nums

nums = [0,1,0,3,12]
result = moveZeroes(nums)
print(result)


## Time Complexity = O(n)
## Space Complexity = O(1)
```

```python
# LeetCode 283 - Move Zeroes
# Pattern: Two Pointer
#
# Problem:
# Move all zeroes to the end of the array
# while maintaining the relative order of non-zero elements.
#
# Example:
# Input  : [0, 1, 0, 3, 12]
# Output : [1, 3, 12, 0, 0]


def moveZeroes(nums):

    # 'left' tells us where the next non-zero
    # element should be placed.
    left = 0

    # 'right' scans the complete array
    # to find non-zero elements.
    for right in range(len(nums)):

        # If right finds a non-zero element
        if nums[right] != 0:

            # Put the non-zero element at the
            # position pointed by left.
            nums[left], nums[right] = nums[right], nums[left]

            # Move left because this position
            # is now correctly occupied by a non-zero.
            left += 1

    # Return the modified array
    return nums


# ---------------- INPUT ----------------

nums = [0, 1, 0, 3, 12]

# Call the function
result = moveZeroes(nums)

# ---------------- OUTPUT ----------------

print(result)

# Output:
# [1, 3, 12, 0, 0]


# ---------------- COMPLEXITY ----------------

# Time Complexity: O(n)
# We scan the array only once using the 'right' pointer.

# Space Complexity: O(1)
# We don't use any extra array.
# We modify the original array in-place.
```
