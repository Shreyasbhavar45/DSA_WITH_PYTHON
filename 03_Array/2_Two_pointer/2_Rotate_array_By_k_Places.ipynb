## Reverse The Array By K places


```python
def rotate(nums, k):

    n = len(nums)

    k = k % n

    def reverse(left, right):

        while left < right:

            nums[left], nums[right] = nums[right], nums[left]

            left += 1
            right -= 1

    reverse(0, n - 1)

    reverse(0, k - 1)

    reverse(k, n - 1)


nums = [1, 2, 3, 4, 5, 6, 7]
k = 3

rotate(nums, k)

print(nums)
```

```python

# LeetCode 189 - Rotate Array
# Pattern: Array + Two Pointer + Reverse
#
# Input:
# nums = [1, 2, 3, 4, 5, 6, 7]
# k = 3
#
# Output:
# [5, 6, 7, 1, 2, 3, 4]


def rotate(nums, k):

    n = len(nums)

    # If k is greater than n, reduce it
    # because rotating n times gives the same array.
    k = k % n

    # Helper function to reverse elements
    # between left and right.
    def reverse(left, right):

        while left < right:

            nums[left], nums[right] = nums[right], nums[left]

            left += 1
            right -= 1

    # Step 1: Reverse the entire array
    #
    # [1, 2, 3, 4, 5, 6, 7]
    # ↓
    # [7, 6, 5, 4, 3, 2, 1]
    reverse(0, n - 1)

    # Step 2: Reverse the first k elements
    #
    # [7, 6, 5 | 4, 3, 2, 1]
    # ↓
    # [5, 6, 7 | 4, 3, 2, 1]
    reverse(0, k - 1)

    # Step 3: Reverse the remaining elements
    #
    # [5, 6, 7 | 4, 3, 2, 1]
    # ↓
    # [5, 6, 7 | 1, 2, 3, 4]
    reverse(k, n - 1)


# ---------------- INPUT ----------------

nums = [1, 2, 3, 4, 5, 6, 7]
k = 3

# Call function
rotate(nums, k)

# ---------------- OUTPUT ----------------

print(nums)

# Output:
# [5, 6, 7, 1, 2, 3, 4]


# ---------------- COMPLEXITY ----------------

# Time Complexity: O(n)
# We reverse the array 3 times.
# O(n) + O(k) + O(n-k) = O(n)

# Space Complexity: O(1)
# We modify the original array in-place.
# No extra array is created.

```
