```python
#Brute Force appraoch

def BruteRotate(nums,k):

    n = len(nums)


    # modulus cha veli jrka vrti lahan ani khali mothi value asli tr tyacha remainder lahan value yete..
    rotations = k % n

    for _ in range(0,rotations):
        e = nums.pop()
        nums.insert(0,e)

    return nums


nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
##better
def BruteRotate(nums,k):

    n = len(nums)

    k = n % k
    nums[:] = nums[n-k:] + nums[:n-k]

    return nums

nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
def rotate(nums,k):
    n = len(nums)

    def reverse(nums,left,right):
        while left < right:
            nums[left],nums[right] = nums[right],nums[left]
            left = left + 1
            right = right - 1

        
    k = k % n
    reverse(nums,n-k,n-1)#reverse lost k elements
    reverse(nums,0,n-k-1)#reverse first  element upto k
    reverse(nums,0,n-1)#reverse all element

    return nums

nums = [1,2,3,4,5,6,7,8]
k = 3
result = rotate(nums,k)
print(result)
```

```python
# Rotate array by one place
def rotateOne(nums):

    n = len(nums)

    temp = nums[n-1]

    for i in range(n-2,-1,-1):
        nums[i+1] = nums [i]
        
    nums[0] = temp

    return nums


nums = [1,2,3,4,5,6]
result = rotateOne(nums)
print("The Rotated Array after One Place is :-",result)
```

## __283.Move Zeroes__


```python
def moveZeroes( nums):
         # left points to the position
    # where the next non-zero element should be placed
        left = 0

        # right checks every element in the array
        for right in range(len(nums)):

            # If current element is non-zero
            if nums[right] != 0:

                # Swap non-zero element with the element at left
                nums[left], nums[right] = nums[right], nums[left]

                # Move left to the next position
                left += 1

        # Return the modified array
        return nums



            
nums = [0,1,0,3,5,6,0]
result = moveZeroes(nums)
print(result)
```

## __LinearSearch__ Concept__



```python
def LinearSearch(nums,target):
    n = len(nums)

    for i in range(n):
        if nums[i] == target:
            return i
        

    return -1


nums = [1,2,3,4,5,6,7]
target = 4
Result = LinearSearch(nums,target)
print("The ELement Found At Index :-",Result)
```

```python
#remove duplicates from sorted array

def Duplicates(nums):


    n = len(nums)

    i = 0
    j = i + 1

    while j < n:

        if nums[j] != nums[i]:

            i+=1

            nums[j],nums[i] = nums[i], nums[j]

        j+=1


    return i+1


nums = [1,1,2,2,3,3,4,4]
result = Duplicates(nums)
print(result)
```

```python
##merge two sorted array without duplicates

class solution(object):
    def MergeTwoSortedArr(self,arr1,arr2):

        result = []
        i = 0 
        j = 0

        while i < len(arr1) and j < len(arr2):

            # comparison

            if arr1[i] <= arr2[j]:

                if len(result) == 0 or result[-1] != arr1[i]:
                    result.append(arr1[i])
                i = i + 1

            else: 

                if len(result) == 0 or result[-1] != arr2[j]:
                    result.append(arr2[j])
                j = j + 1

                
        #if arr1 still has element
        while i < len(arr1):
            if len(result) == 0 or result[-1] != arr1[i]:
                result.append(arr1[i])
            i += 1

        # If arr2 still has elements
        while j < len(arr2):
            if len(result) == 0 or result[-1] != arr2[j]:
                result.append(arr2[j])
            j += 1



        return result
    
arr1 = [1,2,2,4]
arr2 = [2,3,5]

obj = solution()
result = obj.MergeTwoSortedArr(arr1,arr2)
print(result)
```

## __268.Find Missing Number__

```python
#brute force appraoch

def missingNumber(nums):

    n = len(nums)

    for i in range(0,n+1):
        if i not in nums:
            return i
        
    return -1

nums = [3,0,1]
result = missingNumber(nums)
print(result)
```

```python
##optimal approach

def MissingNum(nums):

    n = len(nums)

    expected_sum = 0
    for i in range(n+1):
        expected_sum += i

    actual_sum = 0
    for num in nums:
        actual_sum += num

    return expected_sum - actual_sum


nums = [3,0,1]
result = MissingNum(nums)
print("The Missing Number is :-",result)
```

## __LEETCODE_485 Max CONsecutive Ones__

```python
##
def MaxConsecutiveOnes(nums):
    n = len(nums)

    count = 0
    max_count = 0

    for i in range(n):
        if nums[i] == 1:
            count = count + 1

            if count > max_count:
                max_count = count

        else:
            count = 0

    
    return max_count

nums = [1,0,1,1,1]
result = MaxConsecutiveOnes(nums)
print(result)
```

## __1.TWO_SUM__

```python
#Brute Force appraoch
def twoSum(nums, target):

    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):

            if nums[i] + nums[j] == target:
                return i, j

    return -1, -1


nums = [2,7,11,15]
target = 26
result = twoSum(nums,target)
print(result)
```

```python
def maximum_sum(nums):

    n = len(nums)

    maxi = float('-inf')

    for i in range(n):
        total = 0

        for j in range(i,n):

            total = total + nums[j]

            maxi = max(maxi,total) 

    return maxi

nums = [-2,1,-3,4,-1,2,1,-5,4]

result = maximum_sum(nums)
print(result)
```

```python
##kadane's algorithm
'''"मी current subarray चा sum maintain करतो. जर तो negative झाला तर तो पुढच्या subarray च्या sum ला कमी करेल, 
म्हणून मी तो discard करून पुढून नवीन subarray सुरू करतो. प्रत्येक step ला maximum sum save करत राहतो."

आणि अजून छोटं version:

"Positive sum पुढे carry करायचा, negative sum discard करायचा, आणि प्रत्येक step ला maximum save करायचा."

'''
def subArray(nums):

    n = len(nums)

    total = 0
    maxi = float('-inf')

    for i in range(n):

        total = total + nums[i]

        maxi = max(total,maxi)

        if total < 0:
            total = 0

    return maxi



nums = [-2,1,-3,4,-1,2,1,-5,4]

result = subArray(nums)
print(result)
```

```python
##best time to buy and sell stock\

def best_stock(nums):

    n = len(nums)

    max_profit = float('-inf')

    for i in range(n):

        for j in range(i+1,n):

            current = nums[j] - nums[i]

            if current > max_profit:
                max_profit = current

    return max_profit

nums = [7,2,1,5,6,4,8,9]
result = best_stock(nums)

print(result)
```

```python
def best_stock(price):

    ##edge case
    if not price:
        return 0

    n = len(price)


    minimum_price = price[0]

    max_profit = 0

    for i in range(1,n):

        if price[i] < minimum_price:
            minimum_price = price[i]

        current_price = price[i] - minimum_price
        if current_price > max_profit:
            max_profit = current_price

    return max_profit

price = [7,2,1,5,6,4,8,9]
result = best_stock(price)

print(result)
```
