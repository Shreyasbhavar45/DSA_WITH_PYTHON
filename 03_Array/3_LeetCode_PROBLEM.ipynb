```python
#Brute Force appraoch

def BruteRotate(nums,k):

    n = len(nums)


    # modulus cha veli jrka vrti lahan ani khali mothi value asli tr tyacha remainder lahan value yete..
    rotations = k % n

    for _ in range(0,rotations):
        e = nums.pop()
        nums.insert(0,e)

    return nums


nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
##better
def BruteRotate(nums,k):

    n = len(nums)

    k = n % k
    nums[:] = nums[n-k:] + nums[:n-k]

    return nums

nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
def rotate(nums,k):
    n = len(nums)

    def reverse(nums,left,right):
        while left < right:
            nums[left],nums[right] = nums[right],nums[left]
            left = left + 1
            right = right - 1

        
    k = k % n
    reverse(nums,n-k,n-1)#reverse lost k elements
    reverse(nums,0,n-k-1)#reverse first  element upto k
    reverse(nums,0,n-1)#reverse all element

    return nums

nums = [1,2,3,4,5,6,7,8]
k = 3
result = rotate(nums,k)
print(result)
```

```python
# Rotate array by one place
def rotateOne(nums):

    n = len(nums)

    temp = nums[n-1]

    for i in range(n-2,-1,-1):
        nums[i+1] = nums [i]
        
    nums[0] = temp

    return nums


nums = [1,2,3,4,5,6]
result = rotateOne(nums)
print("The Rotated Array after One Place is :-",result)
```

## __283.Move Zeroes__


```python
def moveZeroes( nums):
         # left points to the position
    # where the next non-zero element should be placed
        left = 0

        # right checks every element in the array
        for right in range(len(nums)):

            # If current element is non-zero
            if nums[right] != 0:

                # Swap non-zero element with the element at left
                nums[left], nums[right] = nums[right], nums[left]

                # Move left to the next position
                left += 1

        # Return the modified array
        return nums



            
nums = [0,1,0,3,5,6,0]
result = moveZeroes(nums)
print(result)
```

## __LinearSearch__ Concept__



```python
def LinearSearch(nums,target):
    n = len(nums)

    for i in range(n):
        if nums[i] == target:
            return i
        

    return -1


nums = [1,2,3,4,5,6,7]
target = 4
Result = LinearSearch(nums,target)
print("The ELement Found At Index :-",Result)
```
