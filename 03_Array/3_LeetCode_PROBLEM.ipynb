```python
#Brute Force appraoch

def BruteRotate(nums,k):

    n = len(nums)


    # modulus cha veli jrka vrti lahan ani khali mothi value asli tr tyacha remainder lahan value yete..
    rotations = k % n

    for _ in range(0,rotations):
        e = nums.pop()
        nums.insert(0,e)

    return nums


nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
##better
def BruteRotate(nums,k):

    n = len(nums)

    k = n % k
    nums[:] = nums[n-k:] + nums[:n-k]

    return nums

nums = [1,2,3,4,5,6]
k = 3
Result = BruteRotate(nums,k)
print(Result)
```

```python
def rotate(nums,k):
    n = len(nums)

    def reverse(nums,left,right):
        while left < right:
            nums[left],nums[right] = nums[right],nums[left]
            left = left + 1
            right = right - 1

        
    k = k % n
    reverse(nums,n-k,n-1)#reverse lost k elements
    reverse(nums,0,n-k-1)#reverse first  element upto k
    reverse(nums,0,n-1)#reverse all element

    return nums

nums = [1,2,3,4,5,6,7,8]
k = 3
result = rotate(nums,k)
print(result)
```

```python
# Rotate array by one place
def rotateOne(nums):

    n = len(nums)

    temp = nums[n-1]

    for i in range(n-2,-1,-1):
        nums[i+1] = nums [i]
        
    nums[0] = temp

    return nums


nums = [1,2,3,4,5,6]
result = rotateOne(nums)
print("The Rotated Array after One Place is :-",result)
```

## __283.Move Zeroes__


```python
def moveZeroes( nums):
         # left points to the position
    # where the next non-zero element should be placed
        left = 0

        # right checks every element in the array
        for right in range(len(nums)):

            # If current element is non-zero
            if nums[right] != 0:

                # Swap non-zero element with the element at left
                nums[left], nums[right] = nums[right], nums[left]

                # Move left to the next position
                left += 1

        # Return the modified array
        return nums



            
nums = [0,1,0,3,5,6,0]
result = moveZeroes(nums)
print(result)
```

```python
#remove duplicates from sorted array

def Duplicates(nums):


    n = len(nums)

    i = 0
    j = i + 1

    while j < n:

        if nums[j] != nums[i]:

            i+=1

            nums[j],nums[i] = nums[i], nums[j]

        j+=1


    return i+1


nums = [1,1,2,2,3,3,4,4]
result = Duplicates(nums)
print(result)
```

```python
##merge two sorted array without duplicates

class solution(object):
    def MergeTwoSortedArr(self,arr1,arr2):

        result = []
        i = 0 
        j = 0

        while i < len(arr1) and j < len(arr2):

            # comparison

            if arr1[i] <= arr2[j]:

                if len(result) == 0 or result[-1] != arr1[i]:
                    result.append(arr1[i])
                i = i + 1

            else: 

                if len(result) == 0 or result[-1] != arr2[j]:
                    result.append(arr2[j])
                j = j + 1

                
        #if arr1 still has element
        while i < len(arr1):
            if len(result) == 0 or result[-1] != arr1[i]:
                result.append(arr1[i])
            i += 1

        # If arr2 still has elements
        while j < len(arr2):
            if len(result) == 0 or result[-1] != arr2[j]:
                result.append(arr2[j])
            j += 1



        return result
    
arr1 = [1,2,2,4]
arr2 = [2,3,5]

obj = solution()
result = obj.MergeTwoSortedArr(arr1,arr2)
print(result)
```

## __268.Find Missing Number__

```python
#brute force appraoch

def missingNumber(nums):

    n = len(nums)

    for i in range(0,n+1):
        if i not in nums:
            return i
        
    return -1

nums = [3,0,1]
result = missingNumber(nums)
print(result)
```

```python
##optimal approach

def MissingNum(nums):

    n = len(nums)

    expected_sum = 0
    for i in range(n+1):
        expected_sum += i

    actual_sum = 0
    for num in nums:
        actual_sum += num

    return expected_sum - actual_sum


nums = [3,0,1]
result = MissingNum(nums)
print("The Missing Number is :-",result)
```

```python
##
def MaxConsecutiveOnes(nums):
    n = len(nums)

    count = 0
    max_count = 0

    for i in range(n):
        if nums[i] == 1:
            count = count + 1

            if count > max_count:
                max_count = count

        else:
            count = 0

    
    return max_count

nums = [1,0,1,1,1]
result = MaxConsecutiveOnes(nums)
print(result)
```

## __1.TWO_SUM__

```python
#Brute Force appraoch
def twoSum(nums, target):

    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):

            if nums[i] + nums[j] == target:
                return i, j

    return -1, -1


nums = [2,7,11,15]
target = 26
result = twoSum(nums,target)
print(result)
```

```python
def Two_sum(nums,target):

    n = len(nums)

    hash_map = {}

    for i in range(n):

        current = nums[i]

        remaining_value = target - current

        if remaining_value in hash_map:
            return [hash_map[remaining_value],i]
        
        hash_map[current] = i

nums = nums = [2,7,11,15]
target = 26
result = Two_sum(nums,target)
print(result)
```

```python
def maximum_sum(nums):

    n = len(nums)

    maxi = float('-inf')

    for i in range(n):
        total = 0

        for j in range(i,n):

            total = total + nums[j]

            maxi = max(maxi,total) 

    return maxi

nums = [-2,1,-3,4,-1,2,1,-5,4]

result = maximum_sum(nums)
print(result)
```

```python
##kadane's algorithm
'''"मी current subarray चा sum maintain करतो. जर तो negative झाला तर तो पुढच्या subarray च्या sum ला कमी करेल, 
म्हणून मी तो discard करून पुढून नवीन subarray सुरू करतो. प्रत्येक step ला maximum sum save करत राहतो."

आणि अजून छोटं version:

"Positive sum पुढे carry करायचा, negative sum discard करायचा, आणि प्रत्येक step ला maximum save करायचा."

'''
def subArray(nums):

    n = len(nums)

    total = 0
    maxi = float('-inf')

    for i in range(n):

        total = total + nums[i]

        maxi = max(total,maxi)

        if total < 0:
            total = 0

    return maxi



nums = [-2,1,-3,4,-1,2,1,-5,4]

result = subArray(nums)
print(result)
```

```python
##best time to buy and sell stock\

def best_stock(nums):

    n = len(nums)

    max_profit = float('-inf')

    for i in range(n):

        for j in range(i+1,n):

            current = nums[j] - nums[i]

            if current > max_profit:
                max_profit = current

    return max_profit

nums = [7,2,1,5,6,4,8,9]
result = best_stock(nums)

print(result)
```

```python
def best_stock(price):

    ##edge case
    if not price:
        return 0

    n = len(price)


    minimum_price = price[0]

    max_profit = 0

    for i in range(1,n):

        if price[i] < minimum_price:
            minimum_price = price[i]

        current_price = price[i] - minimum_price
        if current_price > max_profit:
            max_profit = current_price

    return max_profit

price = [7,2,1,5,6,4,8,9]
result = best_stock(price)

print(result)
```

## __2149.Rearrange Array Elements by Sign__


```python
##code

def rearrange(nums):

    #length of nums
    n = len(nums)

    #positive num stored
    positive_num = []

    #negative num stored
    negative_num = []

    #main answer store in the list
    result = []

    #for iterating each element
    for i in range(n):

        #to store positive element the condition is
        if nums[i] > 0:
            positive_num.append(nums[i])

        #to store negative element the condition is

        else:
            negative_num.append(nums[i])

    #Now we create a two list one for negative and one for negative in presevence order to maintain thier order
    #now we have to add one positive one negative
    #so how should lets see

    #we take two pointer-approach

    positive = 0
    negative = 0

    #we have to iterate positive and negative element up to the last element
    #so

    while positive < len(positive_num):

        #for appending one positive one negativee

        result.append(positive_num[positive])

        #and we movee the pointer

        positive = positive + 1
        

        # for negative element appending

        result.append(negative_num[negative])

        #and here also we move the pointer

        negative = negative + 1


    #now we have to return the result

    return result


nums = [3,1,-2,-5,2,-4]
result = rearrange(nums)
print("The output is for 1st positive and 2nd negative is :-",result)
```

## __128.Longest Consecutive Sequence__


```python
##better approach
def longestConsecutiveSequence(nums):

    if not nums:
        return 0

    nums.sort()

    n = len(nums)

    count = 1
    consecutive = 1

    for i in range(n - 1):

        if nums[i + 1] - nums[i] == 1:
            count = count + 1

            consecutive = max(consecutive,count)


        #skip duplicate element
        elif nums[i + 1] - nums[i] == 0:
            continue

        else:
            count = 1

    return consecutive


nums = [100, 99, 98, 97, 64, 23, 98, 32, 33]

result = longestConsecutiveSequence(nums)
print(result)
```

```python
##optimal approach
def longestConsecutiveSequence(nums):
    n = len(nums)

    my_set = set(nums)
    
    consecutive = 0

    for num in my_set:

        if num - 1 not in my_set:

            x = num

            count = 1

            while x + 1 in my_set:

                count = count + 1

                x = x + 1

            consecutive = max(count,consecutive)

    
    return consecutive

nums = [100,4,200,1,3,2]

result = longestConsecutiveSequence(nums)
print(result)
```

## __Set Matrix Zeroes__


```python
##optimal approch

def setMatrixZeroes(matrix):

    #len of matrix
    row = len(matrix)
    coln = len(matrix[0])

    #0 cha column track
    # Tracker = original 0 kontya row ani column madhe ahe te store karnyasathi
    row_track = [0 for _ in range(row)]#[0,0,0]
    coln_track = [0 for _ in range(coln)]#[0,0,0]


    #for checking where is 0 

    for i in range(row):
        for j in range(coln):

            if matrix[i][j] == 0:
                row_track[i] = -1
                coln_track[j] = -1

    #for making 0

    for i in range(row):
        for j in range(coln):

            if row_track[i] == -1 or coln_track[j] == -1:
                matrix[i][j] = 0

    
    return matrix

    
matrix = [[1,2,3],[4,0,6],[8,9,7]]

result = setMatrixZeroes(matrix)

print(result)

# Time: O(m × n)
# Extra Space: O(m + n)
```
