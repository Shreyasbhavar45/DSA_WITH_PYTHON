## __REVERSE AN ARRAY__

```python
def reverse_array(nums):

    # Start from the first index
    left = 0

    # Start from the last index
    right = len(nums) - 1

    # Keep working until left and right meet
    while left < right:

        # Swap the first and last elements
        nums[left], nums[right] = nums[right], nums[left]

        # Move left pointer one step forward
        left = left + 1

        # Move right pointer one step backward
        right = right - 1

    # Return the reversed array
    return nums


nums = [1, 2, 3, 4, 5]

# Call the function
result = reverse_array(nums)

# Print the reversed array
print(result)
```

```python
def reverse_array(nums):

    left = 0
    right = len(nums)-1

    while left < right:
        
        nums[left],nums[right] = nums[right],nums[left]
        left = left + 1
        right = right - 1

    return nums


nums = [1,2,3,4,5]
result = reverse_array(nums)
print(result)
```

```python
def ispalindrome(nums):

    left = 0
    right = len(nums)-1

    while left < right:
        if nums[left] != nums[right]:
            return False
        
        
        left = left + 1
        right = right - 1
        
    return True
    
        

nums = [1,2,7,4,2,1]
result = ispalindrome(nums)
print(result)
```

```python
##remove duplicates in sorted array...
def duplicate(nums):
    n = len(nums)

    if n == 1:
        return 1
    
    i = 0
    j = i+1 
    while j < n:
        if nums[j] != nums[i]:
            i = i+1

            nums[i],nums[j] = nums[j],nums[i]

        j = j+1

        
    return i + 1

nums = [1,1,2,2,3,3,4,4]
result = duplicate(nums)
print(result)
```

```python
def rotate(nums):
    n = len(nums)

    temp = nums[n-1]

    for i in range(n-2,-1,-1):
        nums[i+1] = nums[i]

    nums[0] = temp

    return nums


nums = [1,3,2,5,6,8]
result = rotate(nums)
print(result)
```
