## __REVERSE AN ARRAY__

```python
def reverse_array(nums):

    # Start from the first index
    left = 0

    # Start from the last index
    right = len(nums) - 1

    # Keep working until left and right meet
    while left < right:

        # Swap the first and last elements
        nums[left], nums[right] = nums[right], nums[left]

        # Move left pointer one step forward
        left = left + 1

        # Move right pointer one step backward
        right = right - 1

    # Return the reversed array
    return nums


nums = [1, 2, 3, 4, 5]

# Call the function
result = reverse_array(nums)

# Print the reversed array
print(result)
```

## __LEETCODE-283 MOVE ZEROES__



```python
def moveZeroes(nums):

    # left points to the position
    # where the next non-zero element should be placed
    left = 0

    # right checks every element in the array
    for right in range(len(nums)):

        # If current element is non-zero
        if nums[right] != 0:

            # Swap non-zero element with the element at left
            nums[left], nums[right] = nums[right], nums[left]

            # Move left to the next position
            left += 1

    # Return the modified array
    return nums


# Normal case
nums = [0, 1, 0, 3, 12]

result = moveZeroes(nums)

print(result)
```
