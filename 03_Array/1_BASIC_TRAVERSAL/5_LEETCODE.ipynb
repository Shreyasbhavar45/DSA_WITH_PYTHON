# Q5 — __Find Missing Number LeetCode 268 — Easy__

```python
## Optimal Appraoch

def missing(nums):
    
    n = len(nums)

    expected_sum = n * (n+1)//2

    actual_sum =  0

    for i in range(n):

        actual_sum = actual_sum + nums[i]

    return expected_sum - actual_sum

nums = [3,0,1]

result = missing(nums)

print(result)


#Time Complexity = O(n) We traverse n elements, and each element is processed once by the for loop, so the number of operations grows linearly with n → O(n).
#Space complexity = O(1)
```
