## __LARGEST ELEMENT IN ARRAY__

```python
#brute force approach

def largest(nums):


    #what is the length or nums
    n = len(nums)


    #we dont know which is the largest so we assume that the first index element is the largest  so we declare
    largest = nums[0]

    #now we have to traverse the whole array to check the largest

    for i in range(n):

        #now we have to give the condition to check the largest if yes we have to update

        if nums[i] >= largest:
            largest = nums[i]

    
    return largest


#function definition

nums = [2,3,7,8,9,9]

#function calling
result = largest(nums)

#print the result
print(result)



### time complexity is O(n) - because we traverse the nums n time

## space complexity is O(1) - because we use only one extra space . and it doesnt use n time

# this is my revision question and i solve it without seeing the codeee or any previous algorithm



```
