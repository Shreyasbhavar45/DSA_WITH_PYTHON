## 
- __PRINT DIAGONAL ELEMENT__

```python
def diagonal(nums):

    rows = len(nums)
    cols = len(nums[0])

    for i in range(0,rows):
        for j in range(0,cols):

            if j == i:
                print(nums[i][j],end = '  ')

            else:
                print('*' , end = '  ')

        print()

    
nums = [[5,20,3],[7,10,9],[1,-52,6]]

result = diagonal(nums)

print(result)
```

```python
def Ddiagonal(nums):

    rows = len(nums)
    cols = len(nums[0])

    for i in range(0,rows):
        for j in range(0,cols):

            if i + j == 2:
                print(nums[i][j],end = '  ')

            else:
                print('*' , end = '  ')

        print()

    
nums = [[5,20,3],[7,10,9],[1,-52,6]]

result = Ddiagonal(nums)

print(result)
```

```python
def lowertriangular(nums):

    rows = len(nums)
    cols = len(nums[0])

    for i in range(0,rows):
        for j in range(0,cols):

            if j >= i:
                print('*',end = ' ')

            else:
                print(nums[i][j], end = ' ')

        print()

    
nums = [[5,20,3],[7,10,9],[1,-52,6]]

result = lowertriangular(nums)

print(result)
```

```python
def uppertriangular(nums):

    rows = len(nums)
    cols = len(nums[0])

    for i in range(0,rows):
        for j in range(0,cols):

            if j >= i:
                print(nums[i][j],end = ' ')

            else:
                print('*' , end = ' ')

        print()

    
nums = [[5,20,3],[7,10,9],[1,-52,6]]

result = uppertriangular(nums)

print(result)
```

```python
##Transpose of matrix


def transpose(nums):

    rows = len(nums)
    cols = len(nums[0])

    #list comprehension
    #expression for variable in iterable
    #[i for i in range(1, 6)]
    # i → expression / output मध्ये काय ठेवायचं
    # for i → loop variable
    # range(1, 6) → कुठून values घ्यायच्या



    result = [[0]*rows for i in range(cols)]

    for i in range(rows):
        for j in range(cols):

            result[j][i] = nums[i][j]

    print(result,end = " ")

nums = [[5,20,3],
        [7,10,9]
        ]

result = transpose(nums)

print(result)
```
