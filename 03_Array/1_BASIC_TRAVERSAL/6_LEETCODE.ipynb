## MAX CONSEXUTIVE ONES


```python
def maxConsecutive(nums):

    n = len(nums)

    count = 0
    max_count = 0

    for i in range(n):

        if nums[i] == 1:
            
            count = count + 1

            if count >= max_count:

                max_count = count
        
        else:
            count = 0

    return max_count



nums = [1,1,0,1,1,1]

result = maxConsecutive(nums)

print(result)

##count → stores the current consecutive streak of 1s. When we see 0, we reset it to 0.
##max_count → stores the longest streak found so far. We never reset it when we see 0.

#Time Complexity = O(n)
#Space Complexity = O(1)
```
