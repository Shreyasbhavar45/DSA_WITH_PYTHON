## __Hashing__

```python
nums = [1,2,2,3,1,1,4]

freq = {}

for num in nums:

    if num in freq:
        freq[num] += 1

    else:
        freq[num] = 1

print(freq)
```

```python
## Hashing Practice : - 2

nums = [4,2,7,2,4,4,9]

# Task : Find the first Number that appears more than once

seen = set()

for num in nums:

    if num in seen:
        print(num)

        break

    else:
        seen.add(num)
```

```python
##find All duplicates
# Find All Duplicates

nums = [4, 2, 7, 2, 4, 4, 9, 7]

seen = set()
duplicates = set()

for num in nums:

    if num in seen:
        duplicates.add(num)

    else:
        seen.add(num)

print(duplicates)
        
```
