## __Hashing__

```python
nums = [1,2,2,3,1,1,4]

freq = {}

for num in nums:

    if num in freq:
        freq[num] += 1

    else:
        freq[num] = 1

print(freq)
```

```python
## Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.
## 217.Contain Duplicate 

def containDuplicate(nums):

    # hashing check element is present or not
    seen = set()

    for num in nums:

        #check number is in freq
        if num in seen:

            return True


        else:

            seen.add(num)

    return False

nums = [1,2,3,1]
result = containDuplicate(nums)
print(result)
```

```python
def intersection(nums1,nums2):

    seen = set()

    for num in nums1:

        if num in seen:

            continue

        else:
            seen.add(num)

    for num in nums2:

        if num in seen:

            continue

        else:
            seen.add(num)
    return seen


nums1 = [1,2,2,1]
nums2 = [2,2]

result = intersection(nums1,nums2)
print(result)

```

```python
## First Duplicate

def duplicate(nums):

    seen = set()

    for num in nums:

        if num in seen:

            return num
        
        else:
            seen.add(num)


nums = [3,5,1,3,7]
result = duplicate(nums)
print(result)
```

## __Frequency Map__

```python
## Frequency Map

def frequency(nums):

    freq = {}
    maximum = 0
    ans = None
    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1

        if freq[num] > maximum:
            maximum = freq[num]
            ans = num

    return ans

nums = [4, 2,7,7,7,7, 2, 4]

result = frequency(nums)
print(result)
```

```python
# Two sum

def twoSum(nums,target):

    seen = {}

    for i in range(len(nums)):

        num = nums[i]

        complement = target - num

        if complement in seen:

            return seen[complement],i
        
        else:

            seen[num] = i

    
nums = [4,6,3,8]
target = 11
result = twoSum(nums,target)
print(result)
```
