```python
def largest_num(arr):
    
    #to find the len of arr
    n = len(arr)#6

    #declare the largest elemnt by assuming
    largest = arr[0]

    #for travesing the array we have using the loop so we can compare each and every element
    for i in range(0,n):#0,5

        #now we put a condition to get largest element by comparing
        if arr[i] > largest:
            largest = arr[i]

    return largest






#function Definition
arr = [12, 5, 8, 21, 3, 17]

#function calling
result = (largest_num(arr))
print(result)
```

```python
def minimum(arr):

    smallest = arr[0]

    for i in range(len(arr)):

        if arr[i] < smallest:
            smallest = arr[i]

    return smallest

arr = [12, 5, 8, 21, 3, 17]
result = minimum(arr)
print(result)
```

```python
def sumOfElement(arr):
     
    total = 0
    for i in range(len(arr)):
        total = total + arr[i]

    return total

arr = [1,2,3,4,5]
result = sumOfElement(arr)
print(result)
```

```python
def countodd(arr): 


    count = 0

    for i in range(len(arr)):
        if arr[i] % 2 != 0:
            count = count + 1
        
    return count


arr = [12,5,8,21,3,27]
result = countodd(arr)
print(result)
```

## __find smallest number in array__

```python
def smallest(nums):

    smallest = nums[0]

    for i in range(len(nums)):
        
        if nums[i] < smallest:
            smallest = nums[i]

    return smallest


nums = [ 1,4,5,4,7,90,8,9,97]
print(smallest(nums))
```

```python
#second smallest
def smallest(nums):

    smallest = float("inf")
    second_smallest = float('inf')

    for i in range(len(nums)):
        
        if nums[i] < smallest:
            second_smallest = smallest
            smallest = nums[i]

        elif nums[i] < second_smallest and nums[i] != smallest:
            second_smallest = nums[i]

    return second_smallest


nums = [ 1,4,5,4,7,90,8,9,97]
print(smallest(nums))
```

```python
#Adjacent Elements pattern
def isSorted(nums):

    n = len(nums)-1

    for i in range(n):
        if nums[i] > nums[i+1]:

            return False
        
    return True

nums = [1,2,8,3,4,5]
print(isSorted(nums))
```

```python
#Adjacent Elements pattern
def isSortedStrictly(nums):
    n = len(nums)-1

    for i in range(n):
        if nums[i] >= nums[i+1]:

            return False
        
    return True

nums = [1,2,2,3,3,4,5,9,10]
print(isSortedStrictly(nums))
```

```python
#diffrence of adjacent variable
def diffrence(nums):

    n = len(nums)-1
    new_diffrence = []

    for i in range(n):
        diff = nums[i+1] - nums[i]
        new_diffrence.append(diff)

    return new_diffrence

nums = [1,23,4,5]
print(diffrence(nums))
```
