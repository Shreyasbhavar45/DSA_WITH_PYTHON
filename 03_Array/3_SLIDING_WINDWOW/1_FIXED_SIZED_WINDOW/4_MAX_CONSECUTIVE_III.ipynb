# __Max Consecutive Ones III__

```python

class Solution(object):
    def longestOnes(self, nums, k):
        left = 0
        zero_count = 0
        max_length = 0

        for right in range(len(nums)):
            if nums[right] == 0:
                zero_count += 1

            while zero_count > k:
                if nums[left] == 0:
                    zero_count -= 1

                left += 1

            current_length = right - left + 1
            max_length = max(max_length, current_length)

        return max_length

nums = [1,1,1,0,0,1,1,1,1]
k = 1
obj = Solution()
print(obj.longestOnes(nums,k))
```

```python

class Solution(object):
    def isValid(self, s):
        stack = []
        pairs = {
            ')': '(',
            '}': '{',
            ']': '['
        }

        for ch in s:
            if ch in pairs:
                if not stack or stack[-1] != pairs[ch]:
                    return False
                stack.pop()
            else:
                stack.append(ch)

        return len(stack) == 0
```
