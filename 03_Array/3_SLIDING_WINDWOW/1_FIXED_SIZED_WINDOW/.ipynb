## __1343. Number of Sub-arrays of Size K and Average Greater than or Equal to Threshold__



```python
## Approach

##Function Definition
def numofsubarray(nums,k,threshold):

    n = len(nums)

    count = 0
    window_sum = 0

    for i in range(k):

        window_sum = window_sum + nums[i]

    average = window_sum // k

    if average >= threshold:

        count = count + 1

    for i in range(k,n):

        window_sum = window_sum + nums[i]

        window_sum = window_sum - nums[i-k]

        average = window_sum // k

        if average >= threshold:

            count = count + 1

    return count


#Driver code
nums = [2,2,2,2,5,5,5,8]
k = 3
threshold = 4

#function Calling
result = numofsubarray(nums,k,threshold)
print(result)
```

```python
## optimal Appraoch

def maxVowels(s,k):

    n = len(s)

    count = 0
    element = 0
    max_count = 0

    for i in range(k):

        element = element + s[i]

        if element in 'aeiou':
            count = count + 1

    max(count,max_count)
       


    for i in range(k,n):

        #adding new element

        element = element + s[i]

        if element in 'aeiou':
            count = count + 1

        #old element
        element = element - s[i-k]

        if element in 'aeiou':
            count = count - 1

    max(count,max_count)

s = 'abciiidef'
k = 3
result = maxVowels(s,k)

print(result)
```
