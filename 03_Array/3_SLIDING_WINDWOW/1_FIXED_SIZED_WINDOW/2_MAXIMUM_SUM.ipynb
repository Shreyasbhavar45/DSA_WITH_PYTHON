## MAXIMUM SUM OF SUBARRAY.


```python
# Codee
def maxSumSubarray(nums, k):

    if k > len(nums):
        return 0

    window_sum = 0
    max_sum = 0

    # First window
    for i in range(k):
        window_sum += nums[i]

    max_sum = window_sum

    # Slide the window
    for i in range(k, len(nums)):

        # Add new element
        window_sum += nums[i]

        # Remove element leaving the window
        window_sum -= nums[i - k]

        # Update maximum
        max_sum = max(max_sum, window_sum)

    return max_sum


nums = [2, 1, 5, 1, 3, 2]
k = 8

print(maxSumSubarray(nums, k))
```
