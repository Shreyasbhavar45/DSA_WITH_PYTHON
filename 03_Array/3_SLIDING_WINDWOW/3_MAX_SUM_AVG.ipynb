##  Leetcode 643 : Maximum Subarray Average I:


```python
## Code

def max_avg(nums,k):


    
    maxi = 0
    total = 0

    for i in range(k):
        total = total + nums[i]

    maxi = total


    for i in range(k,len(nums)):

        total = total + nums[i]

        total = total - nums[i-k]

        maxi = max(total,maxi)

    return maxi/k


nums = [1,12,-5,-6,50,3]
k = 4

result = max_avg(nums,k)
print(result)

```
