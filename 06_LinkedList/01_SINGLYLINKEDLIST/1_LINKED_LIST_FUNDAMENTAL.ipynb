# __1.How To create a Node And Singly Linked List:__

```python
class Node:

    def __init__(self,val):
        # Node Value
        self.val = val
        # Next Node Reference
        self.next = None

# Objects Created
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

# Connecting The Nodes :- Node1 connect to node 2 ...
node1.next = node2
node2.next = node3
node3.next = node4

# print The nodess 
print(node1.val)
print(node1.next.val)


```
