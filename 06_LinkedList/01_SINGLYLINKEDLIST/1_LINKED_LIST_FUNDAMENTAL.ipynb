# __1.How To create a Node And Singly Linked List:__

```python
class Node:

    def __init__(self,val):
        # Node Value
        self.val = val
        # Next Node Reference
        self.next = None

# Objects Created
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

# Connecting The Nodes :- Node1 connect to node 2 ...
node1.next = node2
node2.next = node3
node3.next = node4

# print The nodess 
print(node1.val)
print(node1.next.val)


```

# __2.Create Singly linked list Class__

```python
class Node:

    def __init__(self,val):
        self.val = val
        self.next = None

class SinglyLinkedList:

    def __init__(self):
        self.head = None

    ## Append :- Add Node in Linked list at the end
    def append(self,val):
        new_node = Node(val)
    #what if singly linkedList is empty
        if self.head == None:
            self.head = new_node

    #what if sll is not empty
        else:
            current = self.head
            while current is not None:
                current = current.next
            #how to append
            current.next = new_node
    
    # Time Complexity of append = O(n)
    # space Complexity of append = O(1)





```
