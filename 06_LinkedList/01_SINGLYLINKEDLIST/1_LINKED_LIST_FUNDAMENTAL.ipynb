# __1.How To create a Node And Singly Linked List:__

```python
class Node:

    def __init__(self,val):
        # Node Value
        self.val = val
        # Next Node Reference
        self.next = None

# Objects Created
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

# Connecting The Nodes :- Node1 connect to node 2 ...
node1.next = node2
node2.next = node3
node3.next = node4

# print The nodess 
print(node1.val)
print(node1.next.val)


```

# __2.Create Singly linked list Class__

```python
class Node:

    def __init__(self,val):
        self.val = val
        self.next = None

class SinglyLinkedList:

    def __init__(self):
        self.head = None

    ## Append :- Add Node in Linked list at the end
    def append(self,val):
        new_node = Node(val)
    #what if singly linkedList is empty
        if self.head == None:
            self.head = new_node

    #what if sll is not empty
        else:
            current = self.head
            while current.next is not None:
                current = current.next
            #how to append
            current.next = new_node
    
    # Time Complexity of append = O(n)
    # space Complexity of append = O(1)


    ## How To Traverse The Linked List
    def traversal(self):
        if self.head == None:
            print("Singly Linked List is Empty")

        else:
            current = self.head
            while current is not None:
                print(current.val,end=" ")
                current = current.next

            print()

    ## Insert at specific Position
    def insertion(self,val,position):
        new_node = Node(val)

    # Insert at Beginning
        if position == 0:
            new_node.next = self.head
            self.head = new_node

    # Insert a specific Position
        else:
            current = self.head
            prev_node = None
            count = 0
            while current is not None and count < position:
                prev_node = current
                current = current.next
                count += 1
            
            prev_node.next = new_node
            new_node.next = current

    #Delete a node: Edge cases
    #1) How to delete The head / tell you to delete the head
    def delete(self,val):
        temp = self.head
        if temp.next is not None:
            if temp.val == val:
                self.head = temp.next
                return
            
    # 2) How to delete the specific value.
            else:
                found = False
                prev = None
                while temp is not None:
                    if temp.val == val:
                        found = True
                        break
                    
                    prev = temp
                    temp = temp.next
                
                if found:
                    prev.next = temp.next
                    return

                else:
                    print("Node Not found")
        # del temp
    


sll = SinglyLinkedList()
#append
sll.append(10)
sll.append(20)
sll.append(30)
#insert at beginnig
sll.insertion(5,0)
#insert at specific position
sll.insertion(25,2)
# delete the head
sll.delete(10)
# Traversal
sll.traversal()

## This is ALL About Singly Linked List>>>



```
