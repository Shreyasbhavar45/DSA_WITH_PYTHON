# __Q1.Remove linked list Elements__

```python
# We have Given A Linked List


class Node:

    def __init__(self,val):
        self.val = val
        self.next = None


class SinglyLinkedList:

    def removeLinkedList(self,head,val):
        dummy = Node(0)
        dummy.next = head
        current = head
        prev = dummy
        while current is not None:
            if current.val == val:
                prev.next = current.next

            prev = current
            current = current.next
        return dummy.next

obj = SinglyLinkedList()
head = [1,2,6,3,4,5,6]
val = 6
obj.removeLinkedList(head,val)

```
