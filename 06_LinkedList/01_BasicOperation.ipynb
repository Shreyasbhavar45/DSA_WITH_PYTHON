## __Linked_list__

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

# Insert 5 at the beginning
new_node = Node(5)

new_node.next = head

head = new_node
# Print the linked list
curr = head

while curr is not None:
    print(curr.data)
    curr = curr.next
    
```
