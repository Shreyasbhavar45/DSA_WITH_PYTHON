## __125. Valid Palindrome__

```python
## Optimal Approach

def is_palindrome(s):

    left = 0
    right = len(s)-1

    while left < right:

        ## check is alphabet is numeric or not
        if not s[left].isalnum():
            left = left + 1

            continue

        if not s[right].isalnum():
            right = right - 1

            continue

        if s[left].lower() != s[right].lower():
            return False

    
        left = left + 1
        right = right - 1

    return True

s = "A man, a plan, a canal: Panama"
result = is_palindrome(s)
print(result)

## Time Complexity = O(n)
## Space Complexity = O(1)
```
