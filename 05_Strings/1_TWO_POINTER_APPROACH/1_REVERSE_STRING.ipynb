##  __344. Reverse String__


```python
## Basic String

s = ['h','e','l','l','o']
n = len(s)
for i in (range(n)):
    print(s[i],end = ' ')
```

```python
## Optimal Appraoch

def reverse(s):

    # Len of String
    n = len(s)

    # Two Pointer
    left = 0
    right = n - 1

    # Loop for traverse
    while left < right:

        # Swapping the element
        s[left],s[right] = s[right],s[left]

        # after Swapping move the element
        left = left + 1
        
        right = right - 1

    return s

s = ['h','e','l','l','o']
result = reverse(s)
print(result)
```
