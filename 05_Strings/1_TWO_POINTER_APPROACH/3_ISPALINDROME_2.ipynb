## __IS STRING PALINDROME 2__

```python
def isPalindrome(s):


    def check(left,right):

        while left < right:

            if s[left] != s[right]:

                return False
            

            left += 1
            right -= 1

        return True
    

    left = 0
    right = len(s)-1

    while left < right:

        if s[left] != s[right]:

            return check(left+1,right) or check(left,right+1)
        

        left += 1
        right -= 1

    return True



s = 'abca'

result = isPalindrome(s)
print(result)


### Time Complexity = O(n)
```

```python
class Solution(object):
    def validPalindrome(self, s):

        # This function checks whether the substring
        # between left and right is a palindrome
        def check(left, right):

            # Keep checking until the pointers meet
            while left < right:

                # If characters are different,
                # it is not a palindrome
                if s[left] != s[right]:
                    return False

                # Move left pointer forward
                left += 1

                # Move right pointer backward
                right -= 1

            # All characters matched
            return True


        # Two pointers
        left = 0
        right = len(s) - 1

        # Compare characters from both ends
        while left < right:

            # If characters don't match,
            # we are allowed to delete ONE character
            if s[left] != s[right]:

                # Option 1:
                # Delete the left character
                #
                # Option 2:
                # Delete the right character
                #
                # If either option gives a palindrome → True
                return check(left + 1, right) or check(left, right - 1)

            # Characters matched,
            # so move both pointers
            left += 1
            right -= 1

        # If we reach here, the string is already a palindrome
        # or can be a palindrome without deleting anything
        return True

s1 = Solution()
s = 'abca'
result = s1.validPalindrome(s)

print(result)
# Time Complexity: O(n)
# Space Complexity: O(1)
```
