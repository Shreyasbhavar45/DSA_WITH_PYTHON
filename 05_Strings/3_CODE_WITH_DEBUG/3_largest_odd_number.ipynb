```python
def odd(s):

    # Store the length of the string
    n = len(s)

    # Traverse the string from right to left
    for i in range(n - 1, -1, -1):

        # Convert the current character to integer
        # and check whether the digit is even
        if int(s[i]) % 2 == 0:

            # If even, continue checking the previous digit
            continue

        else:
            # We found the rightmost odd digit
            # Return the string from index 0 to i
            return s[0:i + 1]

    # If there is no odd digit, return an empty string
    return ""


s = "52456"

result = odd(s)

print(result)
```
