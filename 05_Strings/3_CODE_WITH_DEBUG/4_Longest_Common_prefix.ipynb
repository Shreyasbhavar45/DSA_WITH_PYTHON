# __14. Longest Common Prefix__

```python
def longestCommonPrefix(strs):
        if not strs:
            return ""
        
        strs.sort()
        
        left = strs[0]
        right = strs[-1]


        i = 0
        while i < len(left) and i < len(right):
            if left[i]!=right[i]:
                break
            i+=1

        return left[0:i]       


strs = ['flower','flow','flowing']
result = longestCommonPrefix(strs)
print(result)
```
