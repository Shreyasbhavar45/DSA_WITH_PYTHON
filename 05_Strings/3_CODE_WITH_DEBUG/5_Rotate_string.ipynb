## __Rotate String__

```python
def rotateString(s,goal):

    if len(s) != len(goal):

        return False

    else:

        if goal in s+s:
            return True

        else:
            return False


s = 'abcde'
goal = 'abced'
result = rotateString(s,goal)
print(result)


##Time Complexity = O(n)
```

```python
def rotateString(s,goal):

    if len(s) != len(goal):

        return False
    
    return goal in (s+s)


s = 'abcde'
goal = 'cdeab'
result = rotateString(s,goal)
print(result)
```
