```python
#descending order
def selectionSort(arr):
    n = len(arr)

    for i in range(n):
        max_index = i

        for j in range(i+1,n):
            if arr[j] > arr[max_index]:
                max_index = j
            
        #swap
        arr[i],arr[max_index] = arr[max_index],arr[i]

    return arr

arr = [5,3,1,7,2,8,4]
print(selectionSort(arr))

```

```python
#ascending Order
def selectionSort(arr):
    
    #length or arr
    n = len(arr)

    #loop
    for i in range(n):
        min_index = i

        for j in range(i+1,n):
            if arr[j] < arr[min_index]:
                min_index = j

        #swap
        arr[i],arr[min_index] = arr[min_index],arr[i]

    return arr

arr = [5,3,1,7,2,8,4]
print(selectionSort(arr))
```
