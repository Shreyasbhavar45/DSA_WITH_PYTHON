## __153. Find Minimum in Rotated Sorted Array__

```python
##Brute Force Approch

def minimum(nums):

    n = len(nums)

    small = nums[0]

    for i in range(n-1):

        if nums[i] < small:
            small = nums[i]

    return small

nums = [6,1,2,3,4]

result = minimum(nums)

print(result)
```

```python
##Optimal approach

def minimum(nums):

    n = len(nums)

    left = 0 
    right = n - 1

    while left < right:

        mid = left + (right - left)//2

        if nums[mid] > nums[right]:

            left = mid + 1

        else:

            right = mid
        

    return nums[left]

nums = [6,1,2,3,4]

result = minimum(nums)

print(result)
        
```
