```python
##BRUTE FORCE APPROACH

def search(nums, target):


    n = len(nums)


    for i in range(n):

        if nums[i] == target:
            return i

    return -1

#drive code
nums = [4,5,6,7,0,1,2]
#target value
target = 0
#fuction calling
result = search(nums,target)

print("The Element present is index at:",result)
```

```python
# Optimal Approach
# Pattern: Binary Search on Rotated Sorted Array

def Search(nums, target):

    # Get the size of the array
    n = len(nums)

    # Initialize the search range
    left = 0
    right = n - 1

    # Continue searching while the range is valid
    while left <= right:

        # Find the middle index
        mid = left + (right - left) // 2

        # Check if the LEFT half is sorted
        if nums[left] <= nums[mid]:

            # Check if target lies inside the sorted LEFT half
            if nums[left] <= target <= nums[mid]:

                # Target found
                if nums[mid] == target:
                    return mid

                # Target is greater than mid
                # So search in the right part
                elif nums[mid] < target:
                    left = mid + 1

                # Target is smaller than mid
                # So search in the left part
                else:
                    right = mid - 1

            # Target is NOT in the sorted left half
            # So search in the right half
            else:
                left = mid + 1

        # Otherwise, the RIGHT half is sorted
        else:

            # Check if target lies inside the sorted RIGHT half
            if nums[mid] <= target <= nums[right]:

                # Target found
                if nums[mid] == target:
                    return mid

                # Target is greater than mid
                # So search in the right part
                elif nums[mid] < target:
                    left = mid + 1

                # Target is smaller than mid
                # So search in the left part
                else:
                    right = mid - 1

            # Target is NOT in the sorted right half
            # So search in the left half
            else:
                right = mid - 1

    # Target does not exist in the array
    return -1


# Driver Code
nums = [4, 5, 6, 7, 8 ,0, 1, 2]

# Target value we want to find
target = 0

# Function calling
result = Search(nums, target)

# Print the result
print("The Element is present at index:", result)


# Time Complexity: O(log n)
# Space Complexity: O(1)
```
