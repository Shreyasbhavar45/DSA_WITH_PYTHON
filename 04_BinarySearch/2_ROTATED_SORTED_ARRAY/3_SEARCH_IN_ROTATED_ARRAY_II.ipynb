## __🟢 LC 81 — Search in Rotated Sorted Array II__

```python
##Brute Force Appraoch

def search(nums,target):

    n = len(nums)

    for i in range(n):

        if nums[i] == target:

            return True
        
    return False


nums = [10,11,11,12,12,13,13,13,1,2,3,4]
target = 0
result = search(nums,target)

print(result)

##Time Complexity:-O(n)
##Space Complexity:-O(1)
```

```python
## Optimal Approach
def search(nums,target):

    n = len(nums)
    left = 0
    right = n - 1

    while left <= right:

        mid = left + (right - left)//2

        if nums[left] == nums[mid] == nums[right]:

            left = left + 1
            right = right - 1

        
        elif nums[left] <= nums[mid]:

            if nums[left] <= target <= nums[mid]:

                if nums[mid] == target:
                    return True

                elif nums[mid] > target:
                    
                    right = mid - 1
                    
                else:
                    left = mid + 1
                
            else:

                left = mid + 1

        else:

            if nums[mid] <= target <= nums[right]:

                if nums[mid] == target:
                    return True

                elif nums[mid] > target:
                    right = mid - 1
                    
                else:
                    left = mid + 1

            else:
                right = mid - 1

    return False

nums = [10,11,11,12,12,13,13,13,1,2,3,4]
target = 0
result = search(nums,target)

print(result)


#time complexity = O(logn)
#space complexity = O(1)
```

```python
# LeetCode 81
# Search in Rotated Sorted Array II
# Pattern: Binary Search + Rotated Array + Duplicates


def search(nums, target):

    # Array ची length store करतो
    # Example: [10, 11, 12] → n = 3
    n = len(nums)

    # Search space चा starting index
    left = 0

    # Search space चा ending index
    # n - 1 म्हणजे last index
    right = n - 1


    # जोपर्यंत valid search space आहे
    # left <= right म्हणजे अजून elements search करायचे आहेत
    while left <= right:

        # Middle index शोधतो
        # हा overflow avoid करण्यासाठी standard formula आहे
        mid = left + (right - left) // 2


        # ------------------------------------------------
        # CASE 1: DUPLICATES
        # ------------------------------------------------

        # जर left, mid आणि right चे values same असतील,
        # तर कोणता half sorted आहे हे आपण determine करू शकत नाही.
        #
        # Example:
        # [1, 1, 1, 0, 1]
        #  L     M     R
        #
        # nums[left] == nums[mid] == nums[right]
        #
        # म्हणून left आणि right चे duplicate elements
        # safely remove करतो.
        if nums[left] == nums[mid] == nums[right]:

            # Left element discard करतो
            left = left + 1

            # Right element discard करतो
            right = right - 1


        # ------------------------------------------------
        # CASE 2: LEFT HALF IS SORTED
        # ------------------------------------------------

        # जर nums[left] <= nums[mid],
        # तर left ते mid पर्यंतचा भाग sorted आहे.
        #
        # Example:
        # [10, 11, 12, 13, 1, 2, 3]
        #  L         M
        #
        # 10 <= 12 → Left half sorted
        elif nums[left] <= nums[mid]:


            # आता check करतो की target
            # sorted left half मध्ये आहे का.
            #
            # target जर nums[left] आणि nums[mid]
            # यांच्या range मध्ये असेल तर
            # target left half मध्ये असू शकतो.
            if nums[left] <= target <= nums[mid]:


                # जर middle elementच target असेल,
                # तर target सापडला.
                if nums[mid] == target:

                    # Target सापडला म्हणून True return करतो.
                    # Function इथेच stop होतो.
                    return True


                # जर nums[mid] target पेक्षा मोठा असेल,
                # तर target mid च्या left side ला आहे.
                elif nums[mid] > target:

                    # mid ला exclude करून left side search करतो.
                    right = mid - 1


                # जर nums[mid] target पेक्षा छोटा असेल,
                # तर target mid च्या right side ला आहे.
                else:

                    # mid च्या पुढून search करतो.
                    left = mid + 1


            # Target sorted left half च्या range मध्ये नाही.
            else:

                # त्यामुळे target left half मध्ये नसून
                # right side मध्ये असण्याची शक्यता आहे.
                left = mid + 1


        # ------------------------------------------------
        # CASE 3: RIGHT HALF IS SORTED
        # ------------------------------------------------

        # जर left half sorted नसेल,
        # rotated array मुळे right half sorted असतो.
        else:


            # Check करतो की target
            # sorted right half मध्ये आहे का.
            #
            # target जर nums[mid] आणि nums[right]
            # यांच्या range मध्ये असेल तर
            # target right half मध्ये असू शकतो.
            if nums[mid] <= target <= nums[right]:


                # जर mid चाच element target असेल,
                # target सापडला.
                if nums[mid] == target:

                    # Target सापडला → True
                    return True


                # जर nums[mid] target पेक्षा मोठा असेल,
                # target mid च्या left side ला आहे.
                elif nums[mid] > target:

                    # Right side discard करून left ला search करतो.
                    right = mid - 1


                # जर nums[mid] target पेक्षा छोटा असेल,
                # target mid च्या right side ला आहे.
                else:

                    # Left side discard करून right ला search करतो.
                    left = mid + 1


            # Target sorted right half च्या range मध्ये नाही.
            else:

                # म्हणून right half discard करून
                # left side मध्ये search करतो.
                right = mid - 1


    # पूर्ण array search केला पण target सापडला नाही.
    return False


# ------------------------------------------------
# TEST CASE
# ------------------------------------------------

nums = [10, 11, 11, 12, 12, 13, 13, 13, 1, 2, 3, 4]

target = 10

# Function call
result = search(nums, target)

# Answer print करतो
print(result)
```

```python
class Solution(object):
    def search(self, nums, target):

        left = 0
        right = len(nums) - 1

        while left <= right:

            mid = left + (right - left) // 2

            # Target found
            if nums[mid] == target:
                return True

            # Duplicate values → cannot decide which half is sorted
            if nums[left] == nums[mid] == nums[right]:
                left += 1
                right -= 1

            # Left half is sorted
            elif nums[left] <= nums[mid]:

                # Target lies in left sorted half
                if nums[left] <= target < nums[mid]:
                    right = mid - 1

                # Target lies in right half
                else:
                    left = mid + 1

            # Right half is sorted
            else:

                # Target lies in right sorted half
                if nums[mid] < target <= nums[right]:
                    left = mid + 1

                # Target lies in left half
                else:
                    right = mid - 1

        return False
```
