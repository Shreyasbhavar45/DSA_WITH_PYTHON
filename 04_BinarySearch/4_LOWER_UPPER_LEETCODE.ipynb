## Leetcode - 35 Search Insert Position


```python
##Optimal code
##question is - search insert position we have give a array we have to find it index position
# if not insert it and tell the index
# Every element is Unique


###Optimal Approach

#function Definition

def searchPosition(nums,target):

    n = len(nums)

    #position 
    position = n

    #we have need two pointer
    left = 0
    right = n - 1
    
    #now we have to iterate each and every element so we can get our element position

    while left <= right:

        # we have to find mid element

        mid = left + (right - left)//2

        #we have to put the condition to check that element is present on which side or which index
        if nums[mid] >= target:
            position = mid

            right = mid - 1

        else:
            left = mid + 1

    return position



##driver code
nums = [1,3,4,5]
target = 2
##function calling
result = searchPosition(nums,target)

print('The position of the give element is :',result)


#time complexity = O(logn)
#space complexity = o(1)
```
