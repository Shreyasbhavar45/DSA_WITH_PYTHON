## __🔥 Next — LeetCode 162: Find Peak Element__

```python
##Brute Force Appraoch

def peak(nums):

    n = len(nums)

    for i in range(n):

        #First Element
        if i == 0:
            if nums[i] > nums[i+1]:
                return i
            
        #Last Element
        elif i == n - 1:
            if nums[i] > nums[i-1]:
                return i
            
        #Middle element
        else:
            if nums[i] > nums[i-1] and nums[i] > nums[i+1]:
                return i

nums = [1,2,3,4]
result = peak(nums)
print(result)

##Time Complexity = O(n)
##Space Complexity = O(1)
```

```python
def findPeak(nums):

    n = len(nums)

    for i in range(n):

        # first element
        if i == 0:
            if nums[i] > nums[i + 1]:
                return i

        # last element
        elif i == n - 1:
            if nums[i] > nums[i - 1]:
                return i

        # middle element
        else:
            if nums[i] > nums[i - 1] and nums[i] > nums[i + 1]:
                return i
            
nums = [1,2,3,4]
result = findPeak(nums)
print(result)
```

```python
##Optimal Approach
def peakFind(nums):

    n = len(nums)

    left = 0
    right = n - 1

    while left < right:

        mid = left + (right - left)//2

        if nums[mid] < nums[mid+1]:
            left = mid + 1

        else:
            right = mid

    return right

nums = [1,2,3,4]
result = peakFind(nums)
print(result)

#Time Complexity = 0(logn)
#Space Complexity = O(1)
```
