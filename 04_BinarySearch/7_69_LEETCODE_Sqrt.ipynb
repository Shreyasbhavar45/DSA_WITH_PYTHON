### __69_LEETCODE.sqrt__

```python
##function Definition
def sqrt(x):
    left = 0 
    right = x

    while left <= right:

        mid = left + (right - left)//2

        if mid * mid <= x:

            ans = mid
        

            left = mid + 1

        else:
            right = mid - 1

    return ans
 
##driver code
x = 4
##function calling
result = sqrt(x)

print(result)

##time complexity = O(logx)
# - Because in every iteration, binary search removes approximately half of the remaining search space.

##space Complexity = O(1)
# - Because you only use a few variables (left, right, mid, ans) regardless of input size.
```
