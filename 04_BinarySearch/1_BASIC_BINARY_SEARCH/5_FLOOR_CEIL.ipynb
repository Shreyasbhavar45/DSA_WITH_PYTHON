## __FLOOR__

- FLoor means :- largest element in array <= target

## __ceil__

- Ceil means :- Smallest element in array >= target

```python
#floor

def floor(nums,target):

    n = len(nums)

    ceil = -1
    flr = -1

    left = 0
    right = n - 1

    while left <= right:

        mid = left + (right - left)//2

        if nums[mid] <= target:

            flr = nums[mid]

            left = mid + 1

        else:
            right = mid - 1

    return flr 


nums = [2,2,3,4,5,8]
target = 2
result = floor(nums,target)
print(result)
        
```

```python
#ceil
def ceil(nums,target):

    n = len(nums)

    cle = -1

    left = 0
    right = n - 1

    while left <= right:

        mid = left + (right - left)//2

        if nums[mid] >= target:

            cle = nums[mid]

            right = mid - 1

        else:
            left = mid + 1

    return cle


nums = [2,3,4,5,8]
target = 7
result = ceil(nums,target)
print(result)
```

```python
##floor and ceil

def floorAndceil(nums,target):

    ##Len Of Nums
    n = len(nums)

    # IF the value is out of the index return - 1
    flr = -1
    cel = -1

    left = 0
    right = n - 1

    while left <= right:

        ##middle of the nums
        mid = left + (right - left)//2

        #condition

        if nums[mid] <= target:

            flr = nums[mid]

            left = mid + 1

        else:
            right = mid - 1

    left = 0
    right = n - 1

    while left <= right:

        ##middle of the nums
        mid = left + (right - left)//2

        ## condition

        if nums[mid] >= target:

            cel = nums[mid]

            right = mid - 1

        else:
            left = mid + 1


    return flr,cel

nums = [2,3,4,5,8,9]
target = 10
result = floorAndceil(nums,target)
print(result)
```
