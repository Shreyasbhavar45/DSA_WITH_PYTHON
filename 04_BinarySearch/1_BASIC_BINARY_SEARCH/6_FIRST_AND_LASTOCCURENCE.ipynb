## __LEETCODE.35 FIRST AND LAST OCCURENCE__


```python
##brute force


def firstnlast(nums,target):


    n = len(nums)
    first = -1
    last = -1

    for i in range(n):

        if nums[i] == target:
            if first == -1:
                first = i

            last = i

    return first , last

nums = [1,2,3,4,5]

target = 2
result = firstnlast(nums,target)

print(result)
```

```python
##optimal solutionnn


def firstAndLast(nums, target):

    n = len(nums)

    # First occurrence
    left = 0
    right = n - 1
    first = -1

    while left <= right:

        mid = left + (right - left) // 2

        if nums[mid] >= target:

            if nums[mid] == target:

                first = mid

            right = mid - 1

        else:
            left = mid + 1

    # Last occurrence
    left = 0
    right = n - 1
    last = -1

    while left <= right:

        mid = left + (right - left) // 2

        if nums[mid] <= target:

            if nums[mid] == target:

                last = mid

            left = mid + 1

        else:
            right = mid - 1

    return first, last


nums = [1,2,3,4,5,6,7]
target = 2

result = firstAndLast(nums,target)

print(result)


####time complexity = O(logn)
###space complexity = O(1)
```
