##  __LEETCODE :- 875 Koko Eating Bananas__

```python
##Brute Force APPRoach

def Koko(piles, h):

    speed = 1

    while speed <= max(piles):

        total = 0

        # Calculate hours for current speed
        for i in range(len(piles)):

            total += (piles[i] + speed - 1) // speed

        # If all bananas can be eaten within h hours
        if total <= h:
            return speed

        # Try next speed
        speed += 1


piles = [3, 6, 7, 11]
h = 8

result = Koko(piles, h)
print(result)
```

```python
##Optimal Approachh

def koko(piles,h):
    
    left = 1
    right = max(piles)

    while left < right:

        speed = left + (right - left)//2

        total = 0

        for pile in piles:

            total = total + (pile + speed - 1)//speed

        if total <= h:
            right = speed

        else:
            left = speed + 1

    return left

piles = [3,6,7,11]
h = 8

result = koko(piles,h)

print(result)
```

```python
class Solution(object):
    def minEatingSpeed(self, piles, h):

        left = 1
        right = max(piles)

        while left < right:

            speed = left + (right - left) // 2

            total = 0

            # Calculate total hours for this speed
            for pile in piles:
                total += (pile + speed - 1) // speed

            # Speed is enough → try smaller speed
            if total <= h:
                right = speed

            # Speed is not enough → increase speed
            else:
                left = speed + 1

        return left
    
```
