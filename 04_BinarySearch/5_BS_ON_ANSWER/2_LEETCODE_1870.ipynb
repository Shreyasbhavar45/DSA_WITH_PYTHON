## __LeetCode 1870 — Minimum Speed to Arrive on Time is another Binary Search on Answer problem.__

```python
## Brute Force Thinking

def miniSpeed(dist,h):

    n = len(dist)

    speed = 1

    while speed <= max(dist):

        Total = 0

        for i in range(n-1):
            Total = Total + (dist[i]+ speed - 1)//speed
        
        Total = Total + dist[-1]/speed

        if Total <= h:
            return speed
        
        else:
            speed = speed + 1

dist = [1,3,2]
h = 2.7
result = miniSpeed(dist,h)
print(result)
```

```python
def miniSpeed(dist,h):

    n = len(dist)

    left = 1
    right = max(dist)

    while left < right:

        speed = left + (right - left)//2

        Total = 0

        for i in range(n-1):
            Total = Total + (dist[i]+ speed - 1)//speed
        
        Total = Total + dist[-1]/speed

        if Total <= h:
            right = speed
        
        else:
            left = speed + 1

    return right

dist = [1,3,2]
h = 4
result = miniSpeed(dist,h)
print(result)
```

```python
def minSpeedOnTime(dist, hour):

    n = len(dist)
    hour100 = int(round(hour * 100))

    if n > (hour100 + 99) // 100:
        return -1

    left = 1
    right = 10**7

    while left < right:

        speed = left + (right - left) // 2

        total = 0

        for i in range(n - 1):
            total += (dist[i] + speed - 1) // speed

        if total * 100 * speed + dist[-1] * 100 <= hour100 * speed:
            right = speed
        else:
            left = speed + 1

    return left

dist = [1,3,2]
hour = 1.9
result = minSpeedOnTime(dist,hour)
print(result)


## Time Complexity = O(n)*O(log10^7)
## Space Complexity = O(1)
```
