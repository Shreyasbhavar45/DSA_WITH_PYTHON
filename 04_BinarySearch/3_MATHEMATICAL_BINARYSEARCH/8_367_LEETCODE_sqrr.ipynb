## __367.VALID PERFECT SQUARE__

```python
def isPerfectSquare(num):


        left = 0

        right = num

        while left <= right:
                
                mid = left + (right - left)


                if mid * mid == num:
                    
                    return True

                    
                
                elif mid * mid > num:
                       
                       right = mid - 1

                else:
                       
                        left = mid + 1
                    

        return False

num = 10
result = isPerfectSquare(num)
print(result)
```
