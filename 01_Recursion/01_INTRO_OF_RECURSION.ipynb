```python
def printNum(x,n):

    if n == 0:
        return
    
    print(x)

    printNum(x,n-1)

printNum(2,5)
```

```python
#Head Recusrsion..
#print 1 to N using recursion
def func(i,n):
    if i > n:
        return
    print(i)

    #function calling
    func(i+1,n)

func(1,6)
```

```python
#tail recusrion

#print 1 to N using recursion
def func(i,n):
    if i > n:
        return
    
    #function calling
    func(i+1,n)
    print(i)

    

func(1,6)
```

```python
#sum of n number

def func(sum,i,n):
    if i > n:
        print(sum)
        return
    
    func(sum+i,i+1,n)

func(0,1,4)
```

```python
def factorial(num):

    if num == 0 or num == 1:
        return 1
    
    return num * factorial(num-1)

factorial(3)
```

```python
def reverse(arr, left, right):

    # Base Case
    if left >= right:
        return

    # Swap
    arr[left], arr[right] = arr[right], arr[left]

    # Recursive Call
    reverse(arr, left + 1, right - 1)


arr = [10, 20, 30, 40, 50]

reverse(arr, 0, len(arr) - 1)

print(arr)
```

```python
def palindrome(s,left, right):

    #base condition
    if left >= right:
        return True
    
    #condition for palindrome
    if s[left] != s[right]:
        return False
    
    #recusrive call
    return palindrome(s,left+1,right-1)

s = "madam"
print(palindrome(s,0,len(s)-1))
```

```python
def fibonacci(n):

    # Base Cases
    if n == 0:
        
        return 0

    if n == 1:
        return 1

    # Recursive Call
    return fibonacci(n - 1) + fibonacci(n - 2)


n = 6
print(fibonacci(n))

```
